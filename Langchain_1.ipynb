Sequential Workflow using LangChain

```python
import pandas as pd
import numpy as np
```

```python
data=pd.read_excel("/content/My_Financial_Report_Dataset.xlsx")
```

```python
data.head()
```

Chain 1 creates business kpis from the financial dataset provided

```python
from langchain_core.output_parsers import StrOutputParser
```

```python
from langchain_core.prompts import PromptTemplate
```

```python
#prompt for Chain 1
kpi_prompt=PromptTemplate(
    input_variables=["data"],
    template="""
    You are a financial analyst and you have been provided with
    a financial data. Extract the most important financial and business KPIs from the data
    and present them in bullet point format

    financial data:{data}"""
)
```

```python
#creating chain 1
chain1=kpi_prompt | llm | StrOutputParser()
```

```python
#executing chain 1
kpis=chain1.invoke({
    "data":data
})
```
